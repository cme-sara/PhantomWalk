## Testing "hard" systems after DPD initialization
This notebook takes the output of a DPD simulation and uses it as input in a lennard-jones (LJ) simulation that has much stronger repulsions between particles, and bonds that cannot extend beyond the fene_r0 length. 

If the LJ simulation runs, it is evidence that we have a very stable configuration from DPD. It's kindof anticlimactic.

```python
import sys
import os
sys.path.append('../lib/')
import create_system_dpd
from create_system_dpd import create_polymer_system_dpd
from dpd_utils import add_hoomd_writers, run_lj_simulation
import matplotlib
import numpy as np  
import gsd, gsd.hoomd 
import hoomd 
import time
import freud
import matplotlib_inline
import matplotlib.pyplot as plt
%matplotlib inline
matplotlib.style.use("ggplot")
matplotlib_inline.backend_inline.set_matplotlib_formats("svg")
```

```python
num_pol=100
num_mon=100
N = num_pol*num_mon
density=1.4
A=50000
k=50000
r_cut=1.01
min_pair_dist=0.80
seed = 123
dpd_final_frame, closest, s, e_cut = create_polymer_system_dpd(
    num_pol=num_pol,
    num_mon=num_mon,
    density=density,
    A=A,
    k=k,
    r_cut=r_cut,
    bond_l = 0.95,
    gamma=1200,
    sim_seed=seed,
    np_seed=seed,
    write = True,
    min_pair_dist=min_pair_dist,
    energy_scaling = .5,
    bond_tolerance = 0.04,
)
print(f"Finished in time = {s:.2f}s, r_min = {closest:.2f}")
lj = run_lj_simulation(dpd_final_frame=dpd_final_frame,
    fene_r0=1.15,
    fene_delta=0.0,
    random_seed=25,
    dt=0.0005,
    lj_epsilon=1.0,
    lj_sigma=1.0,
    lj_r_cut=2.5,
    fene_k=30,
    fene_epsilon=1.0,
    fene_sigma=1.0,
    angle_k=3.0,
    angle_t0=1.0,
    dihedral_k=3.0,
    dihedral_d=-1,
    dihedral_n=3,
    dihedral_phi0=0
    )
```
